```python
# ============================================================
# URBANFLOW
# Urban Flood Early Warning & Decision Support Prototype
# Google Colab - Single Cell Version
# ============================================================

# Install required packages
!pip -q install folium geopy ipywidgets pandas numpy

# ------------------------------------------------------------
# IMPORTS
# ------------------------------------------------------------

import folium
import pandas as pd
import numpy as np
import ipywidgets as widgets

from IPython.display import display, HTML, clear_output
from geopy.geocoders import Nominatim
from folium.plugins import HeatMap
from google.colab import files

# ------------------------------------------------------------
# URBANFLOW TITLE
# ------------------------------------------------------------

display(HTML("""
<div style="
    background:#111827;
    color:white;
    padding:25px;
    border-radius:15px;
    margin-bottom:20px;
    font-family:Arial;
">
    <h1 style="margin:0;">🌊 UrbanFlow</h1>
    <p style="margin-top:8px;">
        Urban Flood Early Warning & Decision Support System
    </p>
</div>
"""))

display(HTML("""
<h3>📍 Step 1 — Select Your Study Area</h3>
<p>
Search for a city, locality, road or urban area.
The map will move to the selected location.
</p>
"""))

# ------------------------------------------------------------
# LOCATION SEARCH
# ------------------------------------------------------------

location_box = widgets.Text(
    value="",
    placeholder="Example: Velachery, Chennai, India",
    description="Location:",
    layout=widgets.Layout(width="70%")
)

search_button = widgets.Button(
    description="🔍 Find Location",
    button_style="primary"
)

location_output = widgets.Output()

geolocator = Nominatim(
    user_agent="urbanflow_student_project"
)

selected_location = {
    "name": None,
    "latitude": None,
    "longitude": None
}

def search_location(b):

    with location_output:
        clear_output()

        query = location_box.value.strip()

        if not query:
            print("Please enter a location.")
            return

        try:
            result = geolocator.geocode(query, timeout=15)

            if result is None:
                print("❌ Location not found.")
                print("Try a more specific location.")
                return

            selected_location["name"] = result.address
            selected_location["latitude"] = result.latitude
            selected_location["longitude"] = result.longitude

            print("✅ Study area selected:")
            print(result.address)
            print()
            print("Latitude :", round(result.latitude, 6))
            print("Longitude:", round(result.longitude, 6))

        except Exception as e:
            print("Could not connect to the location service.")
            print("Please try again.")

search_button.on_click(search_location)

display(
    widgets.HBox([
        location_box,
        search_button
    ])
)

display(location_output)

# ------------------------------------------------------------
# FILE UPLOAD
# ------------------------------------------------------------

display(HTML("""
<hr>
<h3>📁 Step 2 — Optional Geographic Data Upload</h3>

<p>
You can upload geographic data for your selected study area.
</p>

<ul>
<li>GeoJSON → study-area boundary / roads / drainage / flood points</li>
<li>CSV → latitude and longitude flood locations</li>
</ul>

<p>
If you do not have these files yet, you can continue using the
prototype with estimated/demo values.
</p>
"""))

upload_button = widgets.Button(
    description="📁 Upload Geographic Data"
)

upload_output = widgets.Output()

uploaded_files = []

def upload_data(b):

    with upload_output:

        clear_output()

        result = files.upload()

        uploaded_files.clear()
        uploaded_files.extend(result.keys())

        if uploaded_files:

            print("Uploaded files:")

            for filename in uploaded_files:
                print("  •", filename)

            print()
            print("✅ Files are available to UrbanFlow.")

upload_button.on_click(upload_data)

display(upload_button)
display(upload_output)

# ------------------------------------------------------------
# INPUT PARAMETERS
# ------------------------------------------------------------

display(HTML("""
<hr>
<h3>🌧️ Step 3 — Urban Flood Inputs</h3>
<p>
These controls allow you to test UrbanFlow.
For the current prototype, values that are not connected to
real datasets are clearly treated as estimated/demo inputs.
</p>
"""))

rainfall = widgets.FloatSlider(
    value=50,
    min=0,
    max=300,
    step=5,
    description="🌧️ Rainfall:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="80%"),
    readout_format=".0f"
)

elevation = widgets.Dropdown(
    options=[
        ("Low elevation", 1.0),
        ("Medium elevation", 0.6),
        ("High elevation", 0.2)
    ],
    value=1.0,
    description="📐 Elevation:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="60%")
)

land_use = widgets.Dropdown(
    options=[
        ("Highly urban / impervious", 1.0),
        ("Mixed urban", 0.65),
        ("Green / permeable", 0.25)
    ],
    value=1.0,
    description="🏙️ Land Use:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="60%")
)

drainage = widgets.IntSlider(
    value=50,
    min=0,
    max=100,
    step=5,
    description="🚰 Drainage Capacity:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="80%")
)

historical_floods = widgets.Dropdown(
    options=[
        ("No known historical flooding", 0.0),
        ("Occasional historical flooding", 0.5),
        ("Frequent historical flooding", 1.0)
    ],
    value=1.0,
    description="🌊 Flood History:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="70%")
)

display(rainfall)
display(elevation)
display(land_use)
display(drainage)
display(historical_floods)

# ------------------------------------------------------------
# WHAT-IF PARAMETERS
# ------------------------------------------------------------

display(HTML("""
<h3>🧪 What-If Simulation</h3>
<p>
Test what happens if drainage capacity is improved.
</p>
"""))

improved_drainage = widgets.IntSlider(
    value=80,
    min=0,
    max=100,
    step=5,
    description="Improved Drainage:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="80%")
)

display(improved_drainage)

# ------------------------------------------------------------
# RUN BUTTON
# ------------------------------------------------------------

run_button = widgets.Button(
    description="🤖 RUN URBANFLOW PREDICTION",
    button_style="success",
    layout=widgets.Layout(width="300px", height="45px")
)

result_output = widgets.Output()

display(run_button)
display(result_output)

# ------------------------------------------------------------
# URBANFLOW ENGINE
# ------------------------------------------------------------

def calculate_risk(
    rainfall_value,
    elevation_factor,
    landuse_factor,
    drainage_capacity,
    historical_factor
):

    # Rainfall contribution
    rainfall_score = min(rainfall_value / 150.0, 1.0)

    # Drainage overload
    drainage_load = min(
        (rainfall_value * (0.5 + landuse_factor * 0.5))
        / max(drainage_capacity, 1),
        2.0
    )

    drainage_score = min(drainage_load / 2.0, 1.0)

    # Combine factors
    score = (
        rainfall_score * 0.35 +
        elevation_factor * 0.20 +
        landuse_factor * 0.15 +
        drainage_score * 0.20 +
        historical_factor * 0.10
    )

    score = min(max(score, 0), 1)

    if score < 0.30:
        level = "LOW"
        symbol = "🟢"
    elif score < 0.50:
        level = "MODERATE"
        symbol = "🟡"
    elif score < 0.70:
        level = "HIGH"
        symbol = "🟠"
    else:
        level = "VERY HIGH"
        symbol = "🔴"

    # Estimated water accumulation
    accumulation = (
        rainfall_value
        * (0.4 + landuse_factor * 0.6)
        * (1 - min(drainage_capacity / 120, 0.8))
        * (0.6 + elevation_factor * 0.4)
    )

    accumulation = max(accumulation, 0)

    # Estimated time-to-waterlogging
    if score >= 0.70:
        time_minutes = 20
    elif score >= 0.50:
        time_minutes = 40
    elif score >= 0.30:
        time_minutes = 75
    else:
        time_minutes = 120

    return {
        "score": score,
        "level": level,
        "symbol": symbol,
        "rainfall_score": rainfall_score,
        "drainage_load": drainage_load,
        "accumulation": accumulation,
        "time": time_minutes
    }

# ------------------------------------------------------------
# RUN URBANFLOW
# ------------------------------------------------------------

def run_urbanflow(b):

    with result_output:

        clear_output()

        # Check location
        if selected_location["latitude"] is None:

            display(HTML("""
            <div style="
                background:#fee2e2;
                padding:15px;
                border-radius:10px;
                color:#991b1b;
            ">
            ❌ Please select a study area first.
            </div>
            """))

            return

        lat = selected_location["latitude"]
        lon = selected_location["longitude"]
        name = selected_location["name"]

        # Calculate baseline
        result = calculate_risk(
            rainfall.value,
            elevation.value,
            land_use.value,
            drainage.value,
            historical_floods.value
        )

        # Calculate what-if
        what_if = calculate_risk(
            rainfall.value,
            elevation.value,
            land_use.value,
            improved_drainage.value,
            historical_floods.value
        )

        # ----------------------------------------------------
        # WHY
        # ----------------------------------------------------

        reasons = []

        if rainfall.value >= 100:
            reasons.append("Heavy rainfall")

        elif rainfall.value >= 60:
            reasons.append("Moderate-to-heavy rainfall")

        if elevation.value >= 0.8:
            reasons.append("Low elevation")

        if land_use.value >= 0.8:
            reasons.append("High impervious urban surface")

        if result["drainage_load"] >= 0.7:
            reasons.append("High drainage load")

        if historical_floods.value >= 0.8:
            reasons.append("Frequent historical flooding")

        if not reasons:
            reasons.append("No major high-risk factor detected")

        # ----------------------------------------------------
        # RECOMMENDATION
        # ----------------------------------------------------

        if result["level"] == "VERY HIGH":

            recommendation = """
            Immediately inspect drainage conditions,
            monitor the affected area and prepare traffic
            management or temporary diversion.
            """

        elif result["level"] == "HIGH":

            recommendation = """
            Inspect drainage, check for possible blockage
            and closely monitor the affected road/location.
            """

        elif result["level"] == "MODERATE":

            recommendation = """
            Continue monitoring rainfall and drainage
            conditions in the selected area.
            """

        else:

            recommendation = """
            No immediate high-risk action is suggested.
            Continue normal monitoring.
            """

        # ----------------------------------------------------
        # MAP
        # ----------------------------------------------------

        city_map = folium.Map(
            location=[lat, lon],
            zoom_start=14,
            tiles="OpenStreetMap"
        )

        # Main selected location
        folium.Marker(
            [lat, lon],
            popup=f"""
            <b>UrbanFlow Study Area</b><br>
            {name}<br><br>
            Risk: {result['level']}
            """,
            tooltip="Selected Study Area",
            icon=folium.Icon(
                color="red" if result["score"] >= 0.5 else "green",
                icon="warning-sign"
            )
        ).add_to(city_map)

        # ----------------------------------------------------
        # CREATE DEMO SPATIAL RISK GRID
        # ----------------------------------------------------

        grid_points = []

        offsets = [
            (-0.012, -0.012),
            (-0.012,  0),
            (-0.012,  0.012),
            (0, -0.012),
            (0, 0.012),
            (0.012, -0.012),
            (0.012, 0),
            (0.012, 0.012)
        ]

        for i, (dlat, dlon) in enumerate(offsets):

            local_factor = (
                0.75 +
                0.25 * np.sin(i * 1.7)
            )

            local_score = min(
                max(result["score"] * local_factor, 0),
                1
            )

            if local_score < 0.30:
                color = "green"
                risk = "Low"

            elif local_score < 0.50:
                color = "beige"
                risk = "Moderate"

            elif local_score < 0.70:
                color = "orange"
                risk = "High"

            else:
                color = "red"
                risk = "Very High"

            point_lat = lat + dlat
            point_lon = lon + dlon

            grid_points.append(
                [point_lat, point_lon, local_score]
            )

            folium.CircleMarker(
                location=[point_lat, point_lon],
                radius=13,
                color=color,
                fill=True,
                fill_opacity=0.55,
                popup=f"""
                <b>UrbanFlow Risk Zone</b><br>
                Risk: {risk}<br>
                Risk Score: {local_score:.2f}<br>
                <br>
                <b>Data status:</b><br>
                Prototype / estimated
                """
            ).add_to(city_map)

        # ----------------------------------------------------
        # STUDY AREA CIRCLE
        # ----------------------------------------------------

        folium.Circle(
            location=[lat, lon],
            radius=1500,
            color="blue",
            fill=False,
            popup="UrbanFlow analysis area"
        ).add_to(city_map)

        # ----------------------------------------------------
        # DISPLAY MAP
        # ----------------------------------------------------

        display(HTML("""
        <h2>🗺️ UrbanFlow Risk Map</h2>
        """))

        display(city_map)

        # ----------------------------------------------------
        # RESULTS
        # ----------------------------------------------------

        display(HTML(f"""
        <hr>

        <h2>🤖 UrbanFlow Prediction</h2>

        <div style="
            background:#f3f4f6;
            padding:20px;
            border-radius:12px;
            font-family:Arial;
        ">

        <h3>📍 Study Area</h3>
        <p>{name}</p>

        <h3>{result['symbol']} Risk Level: {result['level']}</h3>

        <p>
        <b>Risk Score:</b> {result['score']:.2f} / 1.00
        </p>

        <p>
        <b>🌧️ Rainfall:</b> {rainfall.value:.0f} mm
        </p>

        <p>
        <b>💧 Estimated Water Accumulation:</b>
        {result['accumulation']:.1f} mm
        </p>

        <p>
        <b>⏱️ Estimated Time-to-Waterlogging:</b>
        {result['time']} minutes
        </p>

        </div>
        """))

        # ----------------------------------------------------
        # WHERE / WHEN / WHY
        # ----------------------------------------------------

        display(HTML(f"""
        <div style="
            display:flex;
            gap:15px;
            flex-wrap:wrap;
            margin-top:20px;
        ">

        <div style="
            flex:1;
            min-width:250px;
            padding:20px;
            border-radius:12px;
            background:#eef2ff;
        ">
        <h3>📍 WHERE?</h3>
        <p>
        The selected study area is identified as the
        UrbanFlow analysis location.
        </p>
        <b>{name}</b>
        </div>

        <div style="
            flex:1;
            min-width:250px;
            padding:20px;
            border-radius:12px;
            background:#fef3c7;
        ">
        <h3>⏱️ WHEN?</h3>
        <p>
        Estimated time-to-waterlogging:
        </p>
        <h2>{result['time']} minutes</h2>
        </div>

        <div style="
            flex:1;
            min-width:250px;
            padding:20px;
            border-radius:12px;
            background:#fee2e2;
        ">
        <h3>❓ WHY?</h3>
        <ul>
        {''.join('<li>'+r+'</li>' for r in reasons)}
        </ul>
        </div>

        </div>
        """))

        # ----------------------------------------------------
        # ALERT
        # ----------------------------------------------------

        if result["level"] in ["HIGH", "VERY HIGH"]:

            alert_background = "#fee2e2"

        elif result["level"] == "MODERATE":

            alert_background = "#fef3c7"

        else:

            alert_background = "#dcfce7"

        display(HTML(f"""
        <div style="
            margin-top:20px;
            padding:25px;
            border-radius:15px;
            background:{alert_background};
        ">

        <h2>🚨 URBANFLOW EARLY WARNING</h2>

        <h3>
        {result['symbol']} {result['level']} FLOOD RISK
        </h3>

        <p>
        <b>Location:</b> {name}
        </p>

        <p>
        <b>Estimated time-to-waterlogging:</b>
        {result['time']} minutes
        </p>

        <p>
        <b>Recommended Action:</b><br>
        {recommendation}
        </p>

        </div>
        """))

        # ----------------------------------------------------
        # WHAT IF
        # ----------------------------------------------------

        improvement = (
            (result["score"] - what_if["score"])
            / max(result["score"], 0.01)
        ) * 100

        display(HTML(f"""
        <hr>

        <h2>🧪 What-If City Simulator</h2>

        <div style="
            display:flex;
            gap:15px;
            flex-wrap:wrap;
        ">

        <div style="
            flex:1;
            min-width:250px;
            padding:20px;
            border-radius:12px;
            background:#f3f4f6;
        ">

        <h3>BEFORE</h3>

        <p>
        Drainage Capacity:
        <b>{drainage.value}</b>
        </p>

        <p>
        Risk:
        <b>{result['level']}</b>
        </p>

        <p>
        Water Accumulation:
        <b>{result['accumulation']:.1f} mm</b>
        </p>

        </div>

        <div style="
            flex:1;
            min-width:250px;
            padding:20px;
            border-radius:12px;
            background:#dcfce7;
        ">

        <h3>AFTER IMPROVEMENT</h3>

        <p>
        Drainage Capacity:
        <b>{improved_drainage.value}</b>
        </p>

        <p>
        Risk:
        <b>{what_if['level']}</b>
        </p>

        <p>
        Water Accumulation:
        <b>{what_if['accumulation']:.1f} mm</b>
        </p>

        </div>

        </div>

        <h3>
        📉 Estimated Risk Reduction:
        {max(improvement,0):.1f}%
        </h3>
        """))

        # ----------------------------------------------------
        # DATA STATUS
        # ----------------------------------------------------

        display(HTML("""
        <hr>

        <h2>📊 Data Status</h2>

        <table style="
            width:100%;
            border-collapse:collapse;
            font-family:Arial;
        ">

        <tr>
            <th style="padding:10px;text-align:left;">
                Input
            </th>

            <th style="padding:10px;text-align:left;">
                Status
            </th>
        </tr>

        <tr>
            <td style="padding:10px;">🌧️ Rainfall</td>
            <td style="padding:10px;">
                🔵 Manual / Demo
            </td>
        </tr>

        <tr>
            <td style="padding:10px;">📐 Elevation / DEM</td>
            <td style="padding:10px;">
                🟡 Estimated
            </td>
        </tr>

        <tr>
            <td style="padding:10px;">🏙️ Land Use</td>
            <td style="padding:10px;">
                🟡 Estimated
            </td>
        </tr>

        <tr>
            <td style="padding:10px;">🛣️ Road Network</td>
            <td style="padding:10px;">
                🟢 Real base map
            </td>
        </tr>

        <tr>
            <td style="padding:10px;">🚰 Drainage Network</td>
            <td style="padding:10px;">
                🟡 Estimated / Uploaded if available
            </td>
        </tr>

        <tr>
            <td style="padding:10px;">🌊 Historical Floods</td>
            <td style="padding:10px;">
                🟡 Estimated / Uploaded if available
            </td>
        </tr>

        </table>
        """))

        # ----------------------------------------------------
        # DISCLAIMER
        # ----------------------------------------------------

        display(HTML("""
        <div style="
            margin-top:25px;
            padding:15px;
            border-radius:10px;
            background:#e5e7eb;
            font-size:13px;
        ">

        <b>Prototype Disclaimer:</b>

        UrbanFlow is a student prototype decision-support system.
        The current version uses manual, estimated and simulated
        inputs where real datasets are not connected.

        Results are intended for demonstration and research purposes
        and must not be treated as an official emergency warning.

        Future versions can connect real-time rainfall,
        DEM, land-use, drainage and historical flood datasets
        and replace the simplified risk model with a trained
        machine-learning model.

        </div>
        """))

run_button.on_click(run_urbanflow)

# ------------------------------------------------------------
# END
# ------------------------------------------------------------

display(HTML("""
<hr>

<div style="
    text-align:center;
    padding:20px;
">

<h3>🌊 UrbanFlow</h3>

<p>
Select an area → Analyze rainfall and urban conditions
→ Identify risk → Generate alert → Test mitigation
</p>

</div>
"""))
```
